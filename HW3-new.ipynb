A **great circle** or orthodrome is the circular intersection of a sphere and a plane passing through the sphere's center point. It displays the ***shortest*** route between two points on the sphere. 

The following website provides a tool to calcualte distance between Latitude/Longitude points on Earth.

**Let's do the caculation using Python**


Compose a program that 

 - assign four floats **lat1, lon1, lat2, and lon2**—(the latitude and longitude, in **degrees**, of two points on the earth)
 - writes the great-circle distance between them using the great-circle distance **d** given by the following equation:

```
d = 6371 arccos[sin(lat1) sin(lat2) + cos(lat1) cos(lat2) cos(lon1 − lon2)]
```

Use your program to compute the great-circle distance between:

- **Paris (48.8575° N, 2.3514° E)**
- **Boise (43.6150° N, 116.2023° W)**.

**Hint:**

- Read references on the math package at: https://docs.python.org/3/library/math.html.
- You need mathematical functions like `math.sin(), math.cos(), math.acos()`.
- The inputs are degrees, whereas Python’s trigonometric functions use **radians**. For example, let's study math.sin(x) function, the reference page https://docs.python.org/3/library/math.html#math.sin metioned: 
"Return the sine of x **radians**". **So, you may need to use `math.radians()` and `math.degrees()` to convert between the two**.


````python
# First, let's difine a function to compute the great circle distance

def great_circle(lat1: float, lon1: float, lat2: float, lon2: float) -> float:
    """
    Calculate the great-circle distance between two points on the Earth's surface.

    Parameters:
    lat1 (float): Latitude of the first point in degrees.
    lon1 (float): Longitude of the first point in degrees.
    lat2 (float): Latitude of the second point in degrees.
    lon2 (float): Longitude of the second point in degrees.

    Returns:
    float: The great-circle distance between the two points in kilometers.
    """
    
    # Convert latitude and longitude from degrees to radians
    lat1 = math.radians(lat1)
    lon1 = math.radians(lon1)
    lat2 = math.radians(lat2)
    lon2 = math.radians(lon2)

    # Calculate the great-circle distance
    d = 6371 * math.acos(math.sin(lat1) * math.sin(lat2) + math.cos(lat1) * math.cos(lat2) * math.cos(lon1 - lon2))
    
    return d
````

````python
# Second, let's computer d between Paris and Boise

lat1 = 48.8575    # Paris
lon1 = 2.3514
lat2 = 43.6150    # Boise
lon2 = -116.2023

d = great_circle(lat1, lon1, lat2, lon2)

# Use F-strings tor formated output
print(f"Latitude and longitude of the first point: ({lat1}, {lon1})")
print(f"Latitude and longitude of the second point: ({lat2}, {lon2})")
print(f"Great-circle distance between the two points: {d:.3f} km")
````

**A Color converter**. Several diﬀerent formats are used to represent color. For example,
the primary format for LCD displays, digital cameras, and web pages, known as the RGB format,
specifies the level of red (R), green (G), and blue (B) on an integer scale from 0 to 255. 

The primary format for publishing books and magazines, known as the CMYK format, specifies the
level of cyan (C), magenta (M), yellow (Y), and black (K) on a real scale from 0.0 to 1.0. Compose
a program that converts RGB to CMYK. 

Take three integers—r, g, and b—from as input and write the equivalent CMYK values as output. If the RGB values are all 0, then the CMY values are all 0 and the K value is 1; otherwise, use these formulas:


\begin{align*}
 w &=\max(r/255, g/255, b/255) \\
 c &=(w-(r/255))/w \\
 m &=(w-(g/255))/w \\
 y &=(w-(b/255))/w \\
 k &=1-w
\end{align*}



````python
def CMYK(r, g, b):
    if r == 0 and g == 0 and b == 0:
        c, m, y, k = 0.0, 0.0, 0.0, 1.0
    else:
        w = max(r / 255, g / 255, b / 255)
        c = (w - r / 255) / w
        m = (w - g / 255) / w
        y = (w - b / 255) / w
        k = 1 - w
    print(f"Cyan level: {c:.3f}")
    print(f"Magenta level: {m:.3f}")
    print(f"Yellow level: {y:.3f}")
    print(f"Black level: {k:.3f}")

# test: if RGB=150,100,30, what would be CMYK
R = 150
G = 100
B = 30

CMYK(R, G, B)
````

**Trigonometric functions.**
Compose a program that compute the sine and cosine functions using their Taylor series expansions 

 
\begin{align*}
 \sin x &= \sum_{n=0}^{\infty}\frac{(-1)^n}{(2n+1)!}x^{2n+1} = x - \frac{x^3}{3!} + \frac{x^5}{5!} - ... \\
 \cos x &= \sum_{n=0}^{\infty}\frac{(-1)^n}{(2n)!}x^{2n} =1 - \frac{x^2}{2!} + \frac{x^4}{4!} - ...
\end{align*}


#### Hint:

- Taylor's series of a function is an infinite sum of terms that are expressed in terms of the function's derivatives at a single point. 
- In numerical calculation, we certainly can't compute sum of **infinite** terms but has to stop when the new term to be added becomes really small, say $$term\leq 1e-15.$$ You can use it for your boolean expression. You can see ``EPSILON" in our example of Newton's method as a reference:

![Accuracy control](Epsilon.png)


````python
EPSILON = 1e-15

def sine(x):
    term = x
    total = 0
    n = 0
    while abs(term) > EPSILON:
        total += term
        n += 1
        term = term * (-x * x) / ((2*n) * (2*n + 1))
    return total

def cosine(x):
    term = 1
    total = 0
    n = 0
    while abs(term) > EPSILON:
        total += term
        n += 1
        term = term * (-x * x) / ((2*n - 1) * (2*n))
    return total


print(sine(0.5))
print(cosine(0.25))
````

**Use Rydberg formula to compute Hydrogen Spectrum.**
In 1888 Johannes Rydberg published his famous formula for calculating the wavelengths $\lambda$ of the emission lines of an hydrogen atom when an atomic electron transit from one orbit to another orbit:

$$
\frac{1}{\lambda}=R\left(\frac{1}{m^2}-\frac{1}{n^2}\right),
$$

where 

- R is the Rydberg constant $R=1.097\times 10^{-2}$ nm$^{-1}$
- $m$ is the principal quantum number of a **low** energy level orbit
- $n$ is the principal quantum number of a **high** energy level orbit
- $n>m$ 


For a given value of $m$, the wavelengths $\lambda$ given by this formula for all $n>m$ forms a series. The first 3 such series, from $m$=1,2, and 3 are known as the Lyman, Balmer and Paschen series after their respective discoverers. 


An example of hydrgen spectrum is:


![Hydrogen Spectrum](Hydrogen_spectrum.svg)


Write a Python program that prints out the wavelengths of the first 5 lines (that means, n=6 if m=1,  n=7 if m=2, n=8, if m=3) in each of these three series. 

You may need use format output (e.g., f-strings or other methods) to orgnize your outputs. 


````python
R = 1.097e-2
names = ["Lyman", "Balmer", "Paschen"]

for m in range(1, 4):
    print(f"\n{names[m-1]} series (m = {m}):")
    print("n\tWavelength (nm)")
    for n in range(m + 1, m + 6):
        lam = 1 / (R * (1/m**2 - 1/n**2))
        print(f"{n}\t{lam:,.2f}")
````
